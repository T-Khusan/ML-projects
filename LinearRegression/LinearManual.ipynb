```python
import numpy as np

def my_train_test_split(X, y, test_size=0.2, random_state=None):

    # Random seed
    if random_state is not None:
        np.random.seed(random_state)

    # Dataset size
    n = len(X)

    # Index yaratish
    indices = np.arange(n)

    # Aralashtirish
    np.random.shuffle(indices)

    # Test dataset soni
    test_count = int(n * test_size)

    # Train/Test indexlar
    test_indices = indices[:test_count]
    train_indices = indices[test_count:]

    # Data ajratish
    x_train = X[train_indices]
    x_test = X[test_indices]

    y_train = y[train_indices]
    y_test = y[test_indices]

    return x_train, x_test, y_train, y_test
```

```python
x_train, x_test, y_train, y_test = my_train_test_split(X, y, test_size = 0.2, random_state=55)
print(x_train.shape, x_test.shape)
print(y_train.shape, y_test.shape)
```

```python
from sklearn.preprocessing import MinMaxScaler

x_scaler = MinMaxScaler()
y_scaler = MinMaxScaler()

x_train = x_scaler.fit_transform(x_train.reshape(-1,1))
x_test = x_scaler.transform(x_test.reshape(-1,1))

y_train = y_scaler.fit_transform(y_train.reshape(-1,1))
y_test = y_scaler.transform(y_test.reshape(-1,1))
```

```python
print(x_train.shape)
```

```python
y_train = y_train.flatten()
y_test = y_test.flatten()
x_train = x_train.flatten()
x_test = x_test.flatten()

```

```python
print(x_train.shape)
```

```python
# Parametrlarga boshlang'ich qiymat berish
b = 0.5  # Boshlang'ich kesish nuqtasi
m_ = 0.5 # Boshlang'ich o'zgarish qiymati
print(f"b: {b}, m: {m_}")
# O'rganish sur'ati
alpha = 0.01
print(f"alpha {alpha}")

# Iteratsiyalar soni
num_iterations = 1000

N = len(x_train)  # namunalar soni
print(f"N {N}")

for i in range(num_iterations):
    Y_hat = b + m_ * x_train  # Bashorat
    error = Y_hat - y_train   # Xatolik
    db = (1/N) * np.sum(error)  # ∂J/∂b
    dm = (1/N) * np.sum(error * x_train)  # ∂J/∂m
    b = b - alpha * db
    m_ = m_ - alpha * dm

    if i % 100 == 0:
        cost = (1/(2*N)) * np.sum(error**2)
        print(f"Iteratsiya {i}: jarima={cost:.2f}, b={b:.2f}, m={m_:.4f}")
```

```python
y_pred_scaled = b + m_ * x_test
```

```python
# Original qiymatga qaytarish
y_pred = y_scaler.inverse_transform(y_pred_scaled.reshape(-1,1))

y_test_original = y_scaler.inverse_transform(y_test.reshape(-1,1))
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
import numpy as np

# MAE
mae = mean_absolute_error(y_test_original, y_pred)

# MSE
mse = mean_squared_error(y_test_original, y_pred)

# RMSE
rmse = np.sqrt(mse)

# R²
r2 = r2_score(y_test_original, y_pred)

print("MAE :", mae)
print("MSE :", mse)
print("RMSE:", rmse)
print("R²  :", r2)
```
